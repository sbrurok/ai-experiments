```python
import pandas as pd
import numpy as np
import plotly.express as px

class Model:
    def __init__(self):
        pass

def load_data():
    # Loading and preprosseses the trainig data
    train_df = pd.read_csv(filepath_or_buffer='data/train.csv')
    X_train = train_df.drop(columns=['price_CHF'])
    # X_train = preprosses(X_train)
    y_train = train_df['price_CHF']
    # y_train = preprosses(y_train)

    # Loading and preprosseses the testing data
    test_df = pd.read_csv(filepath_or_buffer='data/test.csv')
    X_test = test_df
    # X_test = preprosses(X_test)

    return X_train, y_train, X_test

def preprosses(df: pd.DataFrame):

    # creating dummy season values
    if type(df) == pd.Series:
        df = df.fillna(df.mean())
    
    elif type(df) == pd.DataFrame:
        if ('season' in df.columns):
            seasons = df['season'].unique().tolist()

            for s in seasons:
                df[s] = (df['season'] == s).astype(int)

            # removing the unnessesary season column
            df = df.drop(columns=['season'])
    
         # replacing nan values with the column mean
        for col in df.columns:
            df[col] = df[col].fillna(df[col].mean())
    
    return df


X_train, y_train, X_test = load_data()
```
